```python
import tomllib
import subprocess
import json

with open("pyproject.toml", "rb") as f:
    config = tomllib.load(f)

deps = config.get("project", {}).get("dependencies", [])
dev_deps = config.get("dependency-groups", {}).get("dev", [])

all_deps = []
for d in deps + dev_deps:
    # simple parsing of name
    name = d.split("==")[0].split(">=")[0].strip()
    all_deps.append(name)

print("Dependencies in pyproject.toml:")
print(all_deps)

# Run uv pip list to find current version of each
result = subprocess.run(["uv", "pip", "list"], capture_output=True, text=True)
installed = {}
for line in result.stdout.splitlines()[2:]:
    parts = line.split()
    if len(parts) >= 2:
        installed[parts[0].lower()] = parts[1]

# Run uv pip list --outdated to get latest versions
result_outdated = subprocess.run(["uv", "pip", "list", "--outdated"], capture_output=True, text=True)
outdated = {}
for line in result_outdated.stdout.splitlines()[2:]:
    parts = line.split()
    if len(parts) >= 3:
        outdated[parts[0].lower()] = parts[2]

for dep in all_deps:
    dep_lower = dep.lower()
    inst = installed.get(dep_lower, "not installed")
    lat = outdated.get(dep_lower, inst)
    print(f"{dep}: installed={inst}, latest={lat}, status={'outdated' if dep_lower in outdated else 'up-to-date'}")
```

```python
import shutil
import os
import subprocess

git_path = shutil.which("git")
if not git_path:
    # Check typical Windows install paths
    possible_paths = [
        r"C:\Program Files\Git\cmd\git.exe",
        r"C:\Program Files\Git\bin\git.exe",
        r"C:\Program Files (x86)\Git\cmd\git.exe",
        os.path.expandvars(r"%LOCALAPPDATA%\Programs\Git\cmd\git.exe")
    ]
    for p in possible_paths:
        if os.path.exists(p):
            git_path = p
            break

if git_path:
    print(f"Found git at: {git_path}")
    result = subprocess.run([git_path, "status"], capture_output=True, text=True)
    print("STDOUT:")
    print(result.stdout)
    print("STDERR:")
    print(result.stderr)
else:
    print("Git executable not found in common paths.")
```
