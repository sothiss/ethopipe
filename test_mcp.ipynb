```python
import os
import ctypes

root_dir = r"c:\Users\alice\Projects\ethopipe"
found_files = []

# Walk through the directory and find desktop.ini files
for root, dirs, files in os.walk(root_dir):
    for file in files:
        if file.lower() == "desktop.ini":
            full_path = os.path.join(root, file)
            found_files.append(full_path)

print(f"Found {len(found_files)} desktop.ini files:")
for f in found_files:
    print(f)

print("\nDeleting files...")
for f in found_files:
    try:
        # Reset file attributes (remove hidden, system, read-only) before deletion
        ctypes.windll.kernel32.SetFileAttributesW(f, 0x80)
        os.remove(f)
        print(f"Successfully deleted: {f}")
    except Exception as e:
        print(f"Failed to delete {f}: {e}")
```
